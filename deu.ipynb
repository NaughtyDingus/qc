```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

def deutsch_algorithm(oracle_type):
    qc = QuantumCircuit(2, 1)

    qc.x(1)

    qc.h(0)
    qc.h(1)

    if oracle_type == "constant_0":
        pass

    elif oracle_type == "constant_1":
        qc.x(1)

    elif oracle_type == "balanced_x":
        qc.cx(0, 1)

    elif oracle_type == "balanced_notx":
        qc.cx(0, 1)
        qc.x(1)

    else:
        raise ValueError("Invalid oracle type")

    qc.h(0)

    qc.measure(0, 0)

    return qc


oracle = "balanced_x"

qc = deutsch_algorithm(oracle)

print("Deutsch Algorithm Circuit:")
print(qc.draw())

simulator = Aer.get_backend('aer_simulator')
compiled_circuit = transpile(qc, simulator)
job = simulator.run(compiled_circuit, shots=1024)
result = job.result()
counts = result.get_counts()

print("\nMeasurement Counts:", counts)

if '0' in counts:
    print("Result: Function is CONSTANT")
elif '1' in counts:
    print("Result: Function is BALANCED")

plot_histogram(counts)
plt.show()
```
